Ejercicio 16: DataFrame de estudiantes

Enunciado: Crear un DataFrame con información de estudiantes y realizar consultas básicas usando pandas.
Logica: Crear un DataFrame y realizar operaciones básicas con los datos de estudiantes.

```python
import pandas as pd



datos = {
    "Nombre": ["Keren", "Luis", "Carlos", "Suhany", "José",
               "Sofía", "Rolando", "Erick", "Miguel", "Hilda"],
    
    "Carrera": ["Sistemas", "Contaduría", "Derecho", "Medicina", "Arquitectura",
                "Sistemas", "Administración", "Derecho", "Medicina", "Sistemas"],
    
    "Edad": [20, 21, 22, 19, 23, 20, 24, 21, 22, 20],
    
    "Promedio": [85, 78, 90, 88, 70, 95, 80, 76, 89, 92],
    
    "Ciudad": ["San Pedro", "Tegucigalpa", "La Ceiba", "Choloma", "Comayagua",
               "El Progreso", "Santa Rosa", "Danlí", "Tela", "Puerto Cortés"]
}

df = pd.DataFrame(datos)


print("Primeras 5 filas")
print(df.head())


print("\nInformación general")
print(df.info())


print("\nPromedio general:")
print(df["Promedio"].mean())

# Promedio >= 80
print("\nEstudiantes con promedio mayor o igual a 80")
print(df[df["Promedio"] >= 80])

# Mayor a menor
print("\nEstudiantes ordenados por promedio")
print(df.sort_values(by="Promedio", ascending=False))
```

Ejercicio 17: Análisis de ventas con pandas

Enunciado: Crear un DataFrame de ventas y realizar análisis básicos de los datos.
Logica: Crear un DataFrame y analizar las ventas usando pandas.

```python
import pandas as pd

# DataFrame

datos = {
    "Fecha": [
        "2026-01-01","2026-01-02","2026-01-03","2026-01-04","2026-01-05",
        "2026-01-06","2026-01-07","2026-01-08","2026-01-09","2026-01-10",
        "2026-01-11","2026-01-12","2026-01-13","2026-01-14","2026-01-15"
    ],

    "Vendedor": [
        "Jose","Luis","Jefferson","Sheyla","Aaron",
        "Sthefany","Ariana","Carlos","David","Kenny",
        "Hilda","Edwin","Keren","Eliud","Erick"
    ],

    "Producto": [
        "Laptop","Mouse","Teclado","Monitor","Impresora",
        "Fire TV","Alexa","Audifonos","Celular","Ipad",
        "Bateria","Pantalla","Cargador","Camaras","Tablet"
    ],

    "Categoría": [
        "Tecnología","Accesorios","Componentes","Redes","Oficina",
        "Streaming","Hogar Inteligente","Audio","Celulares","Tablets",
        "Energía","Pantallas","Carga","Seguridad","Tecnología"
    ],

    "Cantidad": [2,5,3,1,2,1,4,2,2,1,3,2,5,1,2],

    "PrecioUnitario": [
        25000,500,800,10000,6000,
        1500,1500,800,17000,14000,
        1000,8000,800,7000,4000
    ]
}

df = pd.DataFrame(datos)

# Columna TotalVenta
df["TotalVenta"] = df["Cantidad"] * df["PrecioUnitario"]


print("Total vendido por vendedor")
print(df.groupby("Vendedor")["TotalVenta"].sum())


print("\nProducto más vendido")
print(df.groupby("Producto")["Cantidad"].sum().idxmax())


print("\nCategoría con mayor ingreso")
print(df.groupby("Categoría")["TotalVenta"].sum().idxmax())

# Ventas de mayor a menor
print("\nVentas ordenadas de mayor a menor")
print(df.sort_values(by="TotalVenta", ascending=False))
```

Ejercicio 18: Limpieza de datos con pandas

Enunciado: Crear un DataFrame con datos incompletos y realizar limpieza de datos usando pandas.
Logica: Crear un DataFrame e identificar y limpiar datos nulos.

```python
import pandas as pd

# Datos incompletos

datos = {
    "Nombre": [
        "Keren","Erick","Hilda","Rolando","Aaron",
        "Sthefany","Ariana","Jefferson","Jose","Kennet"
    ],

    "Edad": [
        20,21,None,19,23,
        22,None,24,20,21
    ],

    "Correo": [
        "keren@gmail.com","erick@gmail.com",None,"HilDa@gmail.com","rOland@gmail.com",
        None,"Arianna@gmail.com","JefferL@gmail.com","JoseLop@gmail.com",None
    ],

    "Ciudad": [
        "San Pedro","Tegucigalpa",None,"La Ceiba","Choloma",
        None,"Comayagua","Danlí","Tela",None
    ],

    "Telefono": [
        "9999-1111","8888-2222",None,"7777-3333","6666-4444",
        "5555-5555",None,"4444-6666","3333-7777","2222-8888"
    ]
}

df = pd.DataFrame(datos)

# Valores nulos
print("Valores nulos")
print(df.isnull())

# Cantidad de valores nulos
print("\nValores nulos por columna")
print(df.isnull().sum())

# Ciudades vacías
df["Ciudad"] = df["Ciudad"].fillna("No especificada")

# Eliminar registros sin correo
df = df.dropna(subset=["Correo"])

# DataFrame limpio
print("\nDataFrame limpio")
print(df)
```

Ejercicio 19: Análisis de préstamos

Enunciado: Crear un DataFrame de préstamos y realizar análisis básicos con pandas.
Logica: Crear un DataFrame y analizar las solicitudes de préstamo.

```python
import pandas as pd

# Préstamos

datos = {
    "Cliente": [
        "Jose","Luis","Ana","María","Carlos",
        "Sofía","Pedro","Elena","Miguel","Lucía"
    ],

    "MontoSolicitado": [
        25000,60000,45000,80000,30000,
        70000,55000,40000,90000,65000
    ],

    "Plazo": [
        12,24,18,36,12,
        24,36,18,48,24
    ],

    "IngresoMensual": [
        15000,30000,20000,40000,18000,
        35000,25000,22000,45000,32000
    ],

    "EstadoSolicitud": [
        "Aprobado","Pendiente","Rechazado","Aprobado","Pendiente",
        "Aprobado","Rechazado","Aprobado","Pendiente","Aprobado"
    ]
}

df = pd.DataFrame(datos)


print("Monto promedio solicitado")
print(df["MontoSolicitado"].mean())


print("\nSolicitudes por estado")
print(df["EstadoSolicitud"].value_counts())


print("\nPréstamos aprobados")
print(df[df["EstadoSolicitud"] == "Aprobado"])


print("\nClientes con monto mayor a 50,000")
print(df[df["MontoSolicitado"] > 50000])


print("\nPromedio del monto solicitado por estado")
print(df.groupby("EstadoSolicitud")["MontoSolicitado"].mean())
```

Ejercicio 20: Mini análisis de datos tipo empresa 

Enunciado: Crear un DataFrame de empleados y realizar análisis básicos con pandas.
Logica: Crear un DataFrame y analizar la información de empleados.

```python
import pandas as pd



datos = {
    "Nombre": [
        "Keren","Sheyla","Jose","Rolando","Erick",
        "Hilda","Jefferson","Aaron","Kenny","David"
    ],

    "Departamento": [
        "Tecnología","Recursos Humanos","Ventas","Finanzas","Marketing",
        "Tecnología","Soporte","Ventas","Finanzas","Marketing"
    ],

    "Cargo": [
        "Programador","Asistente","Vendedor","Contador","Diseñador",
        "Analista","Técnico","Supervisor","Auditor","Editor"
    ],

    "Salario": [
        25000,18000,22000,30000,24000,
        28000,20000,26000,32000,23000
    ],

    "AñosExperiencia": [
        2,4,6,8,3,
        10,5,7,12,1
    ],

    "Ciudad": [
        "San Pedro","Tegucigalpa","La Ceiba","Choloma","Comayagua",
        "El Progreso","Tela","Danlí","Puerto Cortés","Santa Rosa"
    ]
}

df = pd.DataFrame(datos)


print("Primeros registros")
print(df.head())


print("\nEstadísticas de salarios")
print(df["Salario"].describe())


print("\nSalario promedio por departamento")
print(df.groupby("Departamento")["Salario"].mean())


print("\nEmpleado con salario más alto")
print(df[df["Salario"] == df["Salario"].max()])


print("\nEmpleados con más de 5 años de experiencia")
print(df[df["AñosExperiencia"] > 5])

print("\nEmpleados ordenados por salario")
print(df.sort_values(by="Salario", ascending=False))

# Columna NivelExperiencia

def nivel(exp):
    if exp <= 3:
        return "Junior"
    elif exp <= 7:
        return "Intermedio"
    else:
        return "Senior"

df["NivelExperiencia"] = df["AñosExperiencia"].apply(nivel)

print("\nDataFrame con nivel de experiencia")
print(df)
```
